```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sl_df = pd.read_csv("./lewidi-ranking-softlabel.csv", sep=";")
sl_df.head(2)
```

```python
pe_df = pd.read_csv("./lewidi-ranking-perspectivist.csv", sep=";")
pe_df.head(2)
```

```python
def asssign_col_method(df):
    df["Method"] = pd.Series(["Others"] * len(df))
    df.loc[df["Team"] == "BoN Appetit Team", "Method"] = "Submission"
    df.loc[df["Team"] == "RLM", "Method"] = "RLM-9b2b"
    return df


pe_df = asssign_col_method(pe_df)
sl_df = asssign_col_method(sl_df)
```

```python
sns.set_context("notebook")

fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(12, 10), constrained_layout=True)
fig.suptitle("LeWiDi Rankings\nLeft: Softlabel, Right: Perspectivist")
datasets = sl_df["Dataset"].unique()
datasets = sorted(datasets)

for row, dataset in enumerate(datasets):
    barplot_kwargs = dict(
        y="Team",
        hue="Method",
        palette={"Others": "C0", "Submission": "C1", "RLM-9b2b": "C2"},
    )
    data = [(sl_df, "WS"), (pe_df, "MAD")]
    for idx, (df, metric) in enumerate(data):
        ax = axes[row, idx]
        data = df.query(f"Dataset == '{dataset}'")
        sns.barplot(data, ax=ax, x=metric, **barplot_kwargs)
        ax.get_legend().remove()
        ax.set_title(dataset)
        ax.set_xlabel(None)
        ax.set_ylabel(None)
        ax.grid(alpha=0.5, axis="x")

# fig.savefig("../imgs/rlm-lewidi-ranking.png", bbox_inches="tight", dpi=300)
```
